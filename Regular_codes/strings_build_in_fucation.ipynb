```python
# Built-in Functions Used With String

```

```python
# 1. len()

#Returns the number of characters.


str1="satej"
print(len(str1))
```

```python
# 2. max()

#Returns the largest character based on character/Unicode ordering.

str1="abcd"
print(max(str1))
```

```python
# 3. min()

#Returns the smallest character.

str1="abcd"

print(min(str1))
```

```python
# 4. sorted()

#Returns characters in sorted order as a list.


str1="python"
print(sorted(str1))
```

```python
#5. type()

#Returns the datatype.

str="satej"

print(type(str))
```

```python
# 6. list()

#Converts string into a list of characters.


Str="Satej"

print(list(Str))
```

```python
# 7. tuple()

#Converts string into tuple.


str="satej"

print(tuple(str))
```

```python
# 8. set()

#Converts string into a set.

#Duplicate characters are removed.

str="apple"

print(set(str))
```

```python
# 9. str()

#Converts a value into a string.

num  = 100
text = str(num)
print(text)
print(type(text))
```

```python
# enumerate()

#Gives index + value.


str1="python"

for index,value in enumerate(str1):
    print(index,value)
```

```python
# 12. reversed()

#Returns a reverse iterator.

str1="apple"

for ch in reversed(str1):
    print(ch,end="")
```

```python
# 13. ord()

#Returns the Unicode code point of a character.

print(ord("A"))
```

```python
# 14. chr()

#Converts a Unicode code point into a character.

print(chr(65))
```

```python
# 15. any()

#Returns True if at least one element is truthy.

#For a string, non-empty characters are truthy.

str1="satej"

print(any(str1))
```

```python
# all()

#Returns True if all elements are truthy.

str="satej"

print(all(str))
```
