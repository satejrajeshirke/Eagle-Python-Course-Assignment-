```python
#String Methods 
```

```python
#1. upper()

#Converts all letters to uppercase
# 
str1="Satej"

print(str1.upper())
```

```python
#2. lower()

#Converts all letters to lowercase.

str1="SATEJ"

print(str1.lower())
```

```python
#3. capitalize()

#Makes the first character uppercase.

name="satej rajeshirke"

print(name.capitalize())
```

```python
# 4. title()

#Makes the first letter of each word uppercase.

str1="satej rajeshirke"

print(str1.title())
```

```python
"""
8. Removing Spaces
6. strip()

Removes spaces from both sides.
"""

str1="           satej         "

print(str1.strip())
```

```python
# 7. lstrip()

#Removes spaces from the left side.

str1="   satej      l"

print(str1.lstrip())
```

```python
# 8. rstrip()

#Removes spaces from the right side.

str1="   satej      "

print(str1.rstrip())
```

```python
# 9. replace()

#Used to replace text.

str2="Java is High level Lang"


str2.replace("Java","python")
```

```python
# 10. split()

#Converts a string into a list.

str1="python java, html"

print(str1.split(","))

```

```python
# 11. join()

#Joins elements into one string.

str1=["satej"," rajeshirke"]

print("".join(str1))
```

```python
#12. find()

#Finds the position of a character or word

str="satej"

print(str.find("a"))
```

```python
#13. index()

#Returns the index position

str1="satej"

print(str.index("j"))
```

```python
# 14. count()

#Counts how many times something occurs.


str1="apple"

print(str1.count("p"))
```

```python
# 5. startswith()

#Checks whether a string starts with particular text.

#Returns True or False.

str1="python"

print(str1.startswith("p"))
```

```python
# 16. endswith()

#Checks whether a string ends with particular text.


str1="python"

print(str1.endswith("n"))
```

```python
# 17. isalpha()

#Checks whether all characters are alphabets.


str1="Satej"

print(str1.isalpha())
```

```python
# 18. isdigit()

#Checks whether all characters are digits.

str1="1234"
print(str1.isdigit())
```

```python
#19. isalnum()

#Checks whether all characters are alphabets or numbers.

str1="satej123"

print(str1.isalnum())
```
