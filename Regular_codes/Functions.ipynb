```python
"""
3. Types of Functions 
1) Built-in Functions 
- print() 
- len() 
- sum() 
- max() 
- min() 
- type() 
- input()

"""

print("Hello")
print(len("helo"))
print(sum([10,20,30]))
print(max([10,20,30]))
print(min([10,20,30,40]))

num=input("enter number")

num=10

type(num)
```

```python
"""
2) User-defined Functions 


"""

def get_data():
    print("data")

get_data()
```

```python
"""
4. Types of User-defined Functions
1) No Arguments, No Return Value 
"""

def geet():
    print("Hello")

geet()
```

```python
"""
2) Arguments, No Return Value 
"""

def greet(name):
    print("Hello",name)

greet("user")
```

```python
"""
3) No Arguments, Return Value

"""
def add():
    return 10+20

result=add()

print(result)
```

```python
""""
4) Arguments and Return Value

"""

def add(num1,num2):
        return num1+num2

add(10,20)
```

```python
"""
5) Return Statement

"""

def square(n):
    return n*n

print(square(5))
```

```python
"""
5. Types of Return 
1) Function without Return - Returns None by default.

"""

def greet():
    print("Good Moring")

greet()
```

```python
"""
2) Function with Single Return Value
"""

def greet(num1,num2):
    return num1+num2

greet(10,20)
```

```python
"""
3)Function with Multiple Return Values
-Python returns them as a tuple.
"""
def add(a,b):
    return a-b,a+b,a*b

add1,sub,mul=add(10,20)

print(add1)
print(sub)
print(mul)
```

```python
"""
6. Function Arguments 
- Arguments are the actual values passed to a function when it is called.

"""

"""
1) Positional Arguments - Arguments are assigned according to their position. The order must be 
the same as the parameters. 
"""

def greet(name,age):
    return name,age

res=greet("demo",23)
print(res)
```

```python
"""
2) Keyword Arguments - Arguments are passed using the parameter name. Order does not 
matter.
"""

def greet(name,age):
    print(name)
    print(age)

greet(age=21,name="satej")
```

```python
""""

3) Default Arguments -
A parameter is assigned a default value. If no argument is passed, the 
default value is used. 

"""

def greet(name="satej"):
        print("hello",name)

greet()
greet("aryan")
```

```python
"""
4) Variable-Length Arguments (*args) -
*args allows a function to accept any number of positional arguments. 
Inside the function, they are stored as a tuple.

"""

def total(*numbers):
    for i in numbers:
        print(i)

total(10,20,30,40)
```

```python
"""
5) Variable-Length Keyword Arguments (**kwargs) -
 **kwargs allows a function to accept any number of keyword 
arguments. Inside the function, they are stored as a dictionary.
"""

def student(**details):
    print(details)

student(name="satej",age=45)
student(name="jay",age=21,city="satara")
```
