```python
"""
primitive DataType


"""
num=10  # Interger
Percenatge=98.8  # float
Is_present=True # boolean
Com=2 + 4j # complex
Not = None #none Type
str1="Satej" #strig

print(num)
print(Percenatge)
print(Is_present)
print(Com)
print(Not)
print(str1)
```

```python
"""
NOn-Primitive DataType
"""

Numbers=[1,2,3,4] #list
Co_ordinates=(10.0,20.0,30.0,60.0) #Tuple
Unique={1,2,3,4,5}  # Set
person={"name":"satej","age":22} #Dict

print(Numbers)
print(Co_ordinates)
print(Unique)
print(person)

```

```python
"""
Type Conversion 
1) Implicit type Conversion
2) Explicit Type Conversion

"""

#1) Implicit Type Conversion

num1=10
num2=12.20

add=num1+num2
print(add)
print(type(add))


#2) Explicit Type

num_str="100"
num2=int(num_str)
print(num2)
print(type(num2))
```
