```python
#list Methods

#Append 
l1=[1,2,3,4,5]
l1.append(30)

print(l1)
```

```python
## remove()

## Removes a specified value.


l1=[1,2,3,4]
l1.remove(4)
print(l1)
```

```python
### pop()

### Removes an element using its index.

l1=[1,2,3,4]
l1.pop(1)
print(l1)
```

```python
###   clear()

## Removes everything

l1=[1,2,3,4]
l1.clear()
print(l1)


```

```python
### sort()

##### Sorts the list.


l1=[1,2,8,6,3,4,5,0]
l2=sorted(l1)
print(l2)

```
