```python
"""
2. Write a function that accepts a list of numbers and returns the largest and smallest number.  
"""


list1=[10,20,30,40,50,60]

def min_max(list1):
    smallest=list1[0]
    highest=list1[0]
    for i in list1:
        if i<smallest:
            smallest=i
        elif i>highest:
            highest=i
        
    return smallest,highest

smallest, highest = min_max(list1)
print("Smallest Number is:", smallest)
print(" highest Number is:",  highest)
```

```python
""""
3. Write a function using  *args to calculate the sum, average, maximum, and minimum of any number of integers.  

"""

def calculation(*args):
    total=sum(args)
    avg=total/len(args)
    maximum=max(args)
    minimum=min(args)

    return total,avg,maximum,minimum

res=calculation(10,20,30,40,50)  # result = (150, 30.0, 50, 10)

print("total is",res[0])
print("average is:",res[1])
print("Maximun :",res[2])
print("minimum:",res[3])     




```

```python
"""
5. Write a function that accepts a string and returns the number of uppercase letters, 
lowercase letters, digits.
"""

str1=str(input("Enter Your Name"))

def check(str1):
    upper_case=0
    lower_case=0
    letter=0
    digits=0
    for i in str1:
        if  i.isupper():
            upper_case+=1
        elif  i.islower():
            lower_case+=1
        elif i.isdigit():
            digits+=1
        

    return upper_case,lower_case, digits,letter

res=check(str1)

print("upper_case",res[0])
print("Lower Case:",res[1])
print("Digites:",res[2])

```

```python
""""

10. Write a function that accepts a sentence and returns the frequency of each word 
using a dictionary. 
"""

sentence=input("Enter A Sentence")


def word_frequency(sentence):
    words=sentence.split()
    frequency={}
    for word in words:
        if word in frequency:
            frequency[word]=frequency[word]+1
        else:
            frequency[word]=1

    return frequency

res=word_frequency(sentence)

print(res)
```

```python
"""
11. Write a function that accepts a list of integers and returns:  
• Sum of all numbers  
• Count of even numbers  
• Count of odd numbers
"""

list1=list(map(int,input("Please Enter numbers with spaces").split()))



def op(list1):
    sum1=0
    count_even=0
    count_odd=0
    for i in list1:
        sum1+=i
        if i%2==0:
            count_even+=1
        else:
            count_odd+=1

    return sum1,count_even,count_odd
res=op(list1)

print("Sum:",res[0])
print("Eeven:",res[1])
print("odd:",res[2])
```
