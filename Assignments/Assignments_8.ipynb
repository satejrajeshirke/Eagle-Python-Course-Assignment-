```python
"""
Write a program to count the frequency of each character in a string using a dictionary.

"""

string1="Hello"


frequency={}

for i in string1:
    if i in frequency:
        frequency[i]=frequency[i]+1
    else:
        frequency[i]=1


print(frequency)
```

```python
"""
Write a program to create a dictionary from two lists: one containing keys and the other containing values.
"""
keys = [1, 2, 3, 4]
values = [10, 20, 30, 40]

dictionary = {}

for i in range(len(keys)):
    dictionary[keys[i]] = values[i]

print(dictionary)
```
