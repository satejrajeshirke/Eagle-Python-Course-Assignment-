```python
# 1. Write a program to count the number of vowels and consonants in a given string. 


name="satej"

vowels=0
consonants=0

for ch in name:
    if ch.lower() in "aeiou":
        vowels+=1
    elif ch.isalpha():
        consonants+=1
    else:
        print("invalid input")


print("Vowels are:",vowels)
print("Consonants are:",consonants)


```

```python
# Write 
#  a given string is a Palindrome. - -
#  Input: madam Output : Palindrome 

name="madam"

for ch in reversed(name):
    print(ch,end="")
```

```python
## Write a program to check whether a given string is a Palindrome. - -
#  Input: madam Output : Palindrome 


name="mada"


if name == name[::-1]:
    print("String is Palindrome")
else:
    print("This is Not Palindrome")


```

```python
#3. Write a program to reverse a string without using slicing ([::-1]) or built-in functions.


str1="satej"

rev=""

for ch in str1:
    rev=ch+rev


print(rev)
```

```python
#4. Write a program to count the occurrence of each character in a string. 
# Input : programming 
# Output:
#  p : 1 
# r : 2
#  o : 1 ... 


str1="programming"
count={}


for ch in str1:
    if ch in count:
        count[ch]+=1
    else:
        count[ch]=1

for ch in count:
    print(ch,";",count[ch])
```

```python
#5. Write a program to remove all duplicate characters from a string while maintaining their original order. 
# - - Input : programming 
# Output : progamin 


str1="programming"

count={}
new=""

for ch in str1:
    if ch not in count:
        count[ch]=1
        new+=ch

print(new)

```

```python
#6. Write a program to find the longest word in a sentence. 
# - Input: Python is an amazing programming language 
#  - Output: programming 

str1="Python is an amazing programming language"


words=str1.split()
longest=""

for word in words:
    if len(word) > len(longest):
        longest=word


print(longest)

```
